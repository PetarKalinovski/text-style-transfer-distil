```python
from pathlib import Path
import pandas as pd
import sys
import json
```

```python
base_qwen_path=Path.cwd().parent/"results" / "qwen3_base_inference_style_transferclaude_judged.json"
with open(base_qwen_path, "r", encoding='utf-8') as f:
    qwen_base_judged = json.load(f)
```

```python
new_qwen_base_judged = []
for item in qwen_base_judged:
    new_qwen_base_judged.append({
        "original" : item["original"],
        "transformed" : item["transformed"],
        "preservation_score" : item["scores"]["preservation"],
        "accuracy_score" : item["scores"]["accuracy"],
        "fluency_score" : item["scores"]["fluency"],
        "reasoning_score" : item["scores"]["reasoning"],
        
    })
```

```python
qwen_base_judged_df=pd.DataFrame(new_qwen_base_judged)
```

```python
qwen_base_judged_df["average_score"] = qwen_base_judged_df[["preservation_score", "accuracy_score", "fluency_score", "reasoning_score"]].mean(axis=1)
```

```python
fine_tune_qwen_path=Path.cwd().parent/"results" / "test_set_inference_resultsclaude_judged.json"
with open(fine_tune_qwen_path, "r", encoding='utf-8') as f:
    qwen_judged = json.load(f)
```

```python
new_qwen_judged = []
for item in qwen_judged:
    new_qwen_judged.append({
        "original" : item["original"],
        "transformed" : item["transformed"],
        "preservation_score" : item["scores"]["preservation"],
        "accuracy_score" : item["scores"]["accuracy"],
        "fluency_score" : item["scores"]["fluency"],
        "reasoning_score" : item["scores"]["reasoning"],
        
    })
```

```python
qwen_judged_df=pd.DataFrame(new_qwen_judged)
```

```python
qwen_judged_df["average_score"] = qwen_judged_df[["preservation_score", "accuracy_score", "fluency_score", "reasoning_score"]].mean(axis=1)
```

```python
qwen_base_judged_df=qwen_base_judged_df.drop(columns=["original", "transformed"])
qwen_judged_df=qwen_judged_df.drop(columns=["original", "transformed"])
```

```python
print("Qwen Base Model Scores:")
print(qwen_base_judged_df.mean())
```

```python
print("Qwen Base Model Scores:")
print(qwen_base_judged_df.mean())
```

```python
print("Fine-tuned Qwen Model Scores:")
print(qwen_judged_df.mean())
```

```python
print("Fine-tuned Qwen Model Scores:")
print(qwen_judged_df.mean())
```
