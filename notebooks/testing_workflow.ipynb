```python
from strands import Agent
from strands_tools import workflow
from strands.models.litellm import LiteLLMModel
import os
```

```python
api_key = os.getenv("OPENROUTER_API_KEY")
os.environ["OPENROUTER_API_KEY"] = api_key
model = LiteLLMModel(
    client_args={
        "api_key": api_key,
    },
    model_id="openrouter/google/gemini-2.5-flash",
    params={
        "max_tokens": 5000,
        "temperature": 0.7,
    }
)
agent=Agent(model=model, tools=[workflow])
```

```python
result= agent.tool.workflow(
        action="status",
        workflow_id="llm_evaluation_workflow"
    )
```

```python
print(result)
```

```python
# First, verify your parent agent works
test_result = agent("Say hello")
print("Parent agent works:", test_result)

# Then try workflow with NO model specified (should inherit)
result = agent.tool.workflow(
    action="create",
    workflow_id="test_parent_model2",
    tasks=[{
        "task_id": "test",
        "description": "Say hello and return the number 5",
        "model_provider": "litellm",
        "model_settings": {
                    "model_id": "openrouter/google/gemini-2.5-flash",
                    "params": {
                        "max_tokens": 500,
                        "temperature": 0.2,
                    },
                    "client_args": {
                        "api_key": api_key,
                    }
                }
    }]
)
print("Create result:", result)

result = agent.tool.workflow(action="start", workflow_id="test_parent_model")
print("Start result:", result)
```
